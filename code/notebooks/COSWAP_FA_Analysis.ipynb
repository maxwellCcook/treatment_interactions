```python
# --- Load COSWAP treatment polygons
fp = join(projdir, 'data/spatial/raw/treatments/ALL_PROJECTS_merge.shp')
coswap_trts = gpd.read_file(fp)
coswap_trts['TRT_TYPE']   = 'Mech Thin'           # all COSWAP units are mechanical thinning
coswap_trts['TRT_ID']     = coswap_trts.index + 1  # unique numeric ID
coswap_trts['raster_key'] = 'mech_thin'            # normalized key — must match catalog scenario
print(f"Treatment units: {len(coswap_trts)}")
print(f"Columns: {list(coswap_trts.columns)}\n")

# --- Load Strategic Focus Areas (analysis extents)
fp = join(projdir, 'data/spatial/raw/boundaries/Strategic_Focus_Areas_COSWAP.shp')
sfas = gpd.read_file(fp)
sfas = sfas[sfas['NAME'].isin(coswap_trts['SFA'].unique())]
print(f"SFAs: {sfas['NAME'].unique()}")

# Dissolve and explode to unique contiguous regions
sfas['dissolve'] = 'Region'
sfas['geometry'] = sfas.geometry.buffer(100)
sfas_d = sfas.dissolve('dissolve', as_index=False).explode(index_parts=False)
print(f"Regions after dissolve/explode: {len(sfas_d)}")
```
